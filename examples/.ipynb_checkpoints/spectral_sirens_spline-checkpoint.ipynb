## Get data from Zenodo

```python
import subprocess
import os

output_dir = './data/'
os.makedirs(output_dir, exist_ok=True)

cmd1 = ["wget", "-c", "-P", output_dir, "https://zenodo.org/records/17143346/files/PE_O5Like_snr20.h5"] # PE file
cmd2 = ["wget", "-c", "-P", output_dir, "https://zenodo.org/records/17143346/files/injections_Ninj_2e7_O5Like_snr20.h5"] # inj file
subprocess.run(cmd1, capture_output=True, text=True)
subprocess.run(cmd2, capture_output=True, text=True)
```

## Load data

```python
import os, sys
os.environ['CHIMERA_ENABLE_GPU'] = "False"
sys.path.append(os.getcwd()+'/../')
from CHIMERA import data as data

file_ev   =  "./data/PE_O5Like_snr20.h5"
theta_pe_det = data.load_gw_pe_samples(file_ev, 
                                 parameters=['m1det', 'm2det', 'dL'], 
                                 return_struct=True)
theta_pe_det = theta_pe_det.update(pe_prior=theta_pe_det.dL**2)
nev = len(theta_pe_det.dL)
print(f'Loaded {nev} GW events')
```

## Instantiate population model

```python
# spline mass model
from CHIMERA.mass.conditioned import pls, setup_spline

num_knots = 10
degree = 3
m_low_prior_lower_bound = 2
m_high_prior_upper_bound = 10
method = 'logspace'
spline_basis_spacing =  'log' 

_, spline_basis_grid, spline_basis = setup_spline(m_low_prior_lower_bound,
    m_high_prior_upper_bound,
    num_knots,
    degree=degree,
    knots_method = method,
    spline_basis_spacing = spline_basis_spacing
)

num_spline_coeffs = spline_basis.shape[1]


# alternatively one can provide user-defined knots
"""
user_defined_knots_list = [ 6.88888889,  8.51851852, 12.07407407, 13.11111111, 16.07407407, 19.62962963,
   22.,         23.03703704, 24.22222222, 28.37037037, 31.77777778, 32.37037037,
   35.33333333, 40.51851852, 42.74074074, 44.66666667, 58.44444444, 63.77777778,
   69.85185185]
user_defined_knots_list += list(np.linspace(user_defined_knots_list[-1], 200,5)[1:])
user_defined_knots_list += list(np.linspace(2, user_defined_knots_list[0],3)[:-1])
knots = np.sort(user_defined_knots_list)
num_knots = len(knots)
  _, spline_basis_grid, spline_basis = setup_spline(m_low_prior_lower_bound,
    m_high_prior_upper_bound,
    num_knots,
    degree=degree,
    inner_knots=knots)
"""

mass = pls(m_low=5.1, m_high=87, alpha=3.4, beta = 1.1, spline_basis_grid=spline_basis_grid, spline_basis=spline_basis)
```

```python
## cosmo and rate and full population
from CHIMERA.cosmo import flrw
from CHIMERA.rate import madau_dickinson
from CHIMERA import population

cosmo = flrw(H0 = 70., 
             Om0=0.25, 
             z_max = 5.)
rate = madau_dickinson(gamma = 2.7,
                       kappa =  3.,
                       zp = 2.)


population = population(cosmo, mass, rate, scale_free=True)
```

## Instantiate the Selection Function module

```python
from CHIMERA import selection_function

file_inj = "./data/injections_Ninj_2e7_O5Like_snr20.h5"
theta_inj_det = data.load_injection_data(file_inj, snr_cut=20, frame='source')

sel_fcn = selection_function(theta_inj_det, N_inj=20*1e6)
```

## Instantiate the Hyperlike

```python
from CHIMERA import hyperlikelihood

hyperlike = hyperlikelihood(  # data
  theta_gw_det = theta_pe_det,
  # population and grids
  population=population,
  # integration grid resolution
  z_grids_res = 300,  
  # selection function
  selection_function=sel_fcn,
  # numerical stability
  pe_neff = 2.0,
  inj_neff = None, # default to 5*Nev  
  # KDE settings
  kind_kde = 'binned',
  kernel = 'epan',
  kde_bw = None, # default to scott
  num_bins = 200,
)
```

## 1d on H0

### Note that 1d posterior differs from the one obtained using the plp notebook since in this notebook we modeled the mass as a PLS in which the spline coefficients are all zero, implying that the actual mass model is a simple truncated and smoothed power law with no additional feautures

```python
import numpy as np
import tqdm
import matplotlib.pyplot as plt

H0 = np.linspace(50, 90, 50)
log_like_evs_H0 = np.zeros((H0.shape[0], hyperlike.nevents))
loglike_H0 = np.zeros_like(H0)
N_exp_H0 = np.zeros_like(H0)

for i, h0 in tqdm.tqdm(enumerate(H0)):
  log_like_evs, N_exp, neff_inj, log_hyperlike = hyperlike.compute_all(H0=h0)
  log_like_evs_H0[i] = log_like_evs
  N_exp_H0[i] = N_exp
  loglike_H0[i] = log_hyperlike

# exponentiate and normalize
log_like_evs_H0 -= np.nanmax(log_like_evs_H0)
like_evs_H0 = np.exp(log_like_evs_H0) 
norms_evs = np.trapz(like_evs_H0, x=H0, axis=0)

loglike_H0 -= np.nanmax(loglike_H0)
like_H0 = np.exp(loglike_H0)
like_H0 /= np.trapz(like_H0, H0)

for e in range(hyperlike.nevents):
    lab = 'single event posteriors' if e == 0 else None
    plt.plot(H0, like_evs_H0[:,e]/norms_evs[e], c='gray', lw=0.75, alpha=0.75, label = lab )
plt.plot(H0, like_H0, label='posterior')
plt.axvline(70, ls='--', c='k', label= 'fiducial')
plt.legend()
plt.show()

```
