```python
import os, sys
os.environ['CHIMERA_ENABLE_GPU'] = "False"
sys.path.append(os.getcwd()+'/../')
import jax.numpy as jnp
import matplotlib.pyplot as plt
from CHIMERA.population.mass.paired import neural_density
from CHIMERA.population.mass import p_m1m2, p_m1, pdf_joint_and_marg

model = neural_density(hidden_size=4, depth=2)
print("Z (2D norm) at fiducial:", model.norm_2d)
model
```

```python
import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt

def sample_random_params(input_size, hidden_size, depth, key, scale=0.8):
  sizes = [input_size] + [hidden_size] * depth
  keys = jax.random.split(key, depth + 1)
  Ws, bs = [], []
  for i in range(depth):
    w_key, b_key = jax.random.split(keys[i])
    Ws.append(scale * jax.random.normal(w_key, (sizes[i + 1], sizes[i])))
    bs.append(scale * jax.random.normal(b_key, (sizes[i + 1],)))

  Ws.append(scale * jax.random.normal(keys[-1], (1, arch.hidden_size)))
  return Ws, bs

# ---------------------------------------------------------------------------
# A few random realizations, overlaid on the marginal
# ---------------------------------------------------------------------------

n_realizations = 5
master_key = jax.random.PRNGKey(1)
keys = jax.random.split(master_key, n_realizations)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

grid_for_joint = None  # keep the last realization's grid for the joint panel

for i, key in enumerate(keys):
  Ws, bs = sample_random_params(model, key, scale=0.8)
  model_rand = model.update(Ws=Ws, bs=bs)

  grid = pdf_joint_and_marg(model_rand, res=(4000, 2000))
  grid_for_joint = grid

  norm_m1 = jnp.trapezoid(grid['p_m1_marg'], grid['m1'])
  norm_m2 = jnp.trapezoid(grid['p_m2_marg'], grid['m2'])
  print(f"realization {i}: int p(m1) dm1 ~ {norm_m1:.4f}, "
        f"int p(m2) dm2 ~ {norm_m2:.4f}")

  axes[0].plot(grid['m1'], grid['p_m1_marg'], alpha=0.8, label=f"draw {i}")

axes[0].set_yscale('log')
axes[0].set_ylim(1e-5, 1)
axes[0].set(xlabel=r'$m_1\ [M_\odot]$', ylabel=r'$p(m_1)$',
            title='Marginal $p(m_1)$, random realizations')
axes[0].legend(fontsize=8)

im = axes[1].pcolormesh(
  grid_for_joint['m1mesh'], grid_for_joint['m2mesh'],
  grid_for_joint['p_joint'], shading='auto',
)
axes[1].set(xlabel='$m_1$', ylabel='$m_2$',
            title='Joint $p(m_1,m_2)$, last realization')
fig.colorbar(im, ax=axes[1])

fig.tight_layout()
plt.show()
```
